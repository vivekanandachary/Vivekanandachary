```python
#billing system


item1 = input("Enter the item name: ")
item1_price = float(input("Enter the price of item 1: "))

item2 = input("Enter the item name: ")
item2_price = float(input("Enter the price of item 2: "))

item3 = input("Enter the item name: ")
item3_price = float(input("Enter the price of item 3: "))

print(item1, item1_price)
print(item2, item2_price)
print(item3, item3_price)

total = item1_price + item2_price + item3_price
print("Total price:", total)

discount = total * 0.10

after_discount = total - discount
print("Amount to pay:", after_discount)
```

```python
#Sets and Dictionaries, Functions
#SETS
# 1. Defining a set
students = {"Dhanu", "Prudhvi", "Mohan", "Sahil"}
print(type(students))  # Output: <class 'set'>

# 2. Duplicate handling (Duplicates are automatically removed)
marks = {"30", "40", "60", "30", "40"}
print(marks)


# 3. Attempting to index a set results in an error
#print(students[2])


# 4. Adding a new unique element
students.add("shashi")
print(students)


# 5. Adding an existing element (Duplicate)
students.add("shashi")
print(students)

# 6. To add multiple names
students.update(["Ravi", "shiva", "lakshay"])
print(students)

# 7. To remove a name
students.remove("Ravi")
print(students)
```

```python

Ai_students = {"vivek", "tarun", "abhi", "galiv", "sameer"}
El_students = {"abhi", "shiva", "shashi", "Ravi"}

# 1. Intersection (&)
y = Ai_students & El_students
print("Intersection (y):", y)


# 2. Union (|)
x = Ai_students | El_students
print("Union (x):", x)


# 3. Difference (-)
z = Ai_students - El_students
print("Difference (z):", z)
```
